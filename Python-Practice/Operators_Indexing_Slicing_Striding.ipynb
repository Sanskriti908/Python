# Indexing , Slicing , Striding

## Indexing

```python
name = 'VICTORIA'
name [0] 
```

```python
name[-1]
```

## Slicing

```python
# Top 5 alphabets
name[0:5] #or name[:5]
```

```python
name[:0]
```

```python
name[7:]
```

```python
name[7:8]
```

```python
name[7:0]
```

```python
# bottom 5
name[-5:]
```

```python
name[-5:-1]
```

```python
name[-5:0]
```

# Striding

```python
# with positive indexing and negative jump
name[::-1]
```

```python
name[-8:-5:-1]
```

```python
name[-5:-8:-1]
```

```python
name[-5::-1]
```

```python
name[-8:-5]
```
