# 1. IF CONDITION

```python
age = 14
if age >= 18:
    print('Eligible to vote')
```

```python
age = 24
if age >= 18:
    print('Eligible to vote')
```

## 2.IF-ELSE 

```python
age = 14
if age >= 18:
    print('Eligible to vote')
else:
    print('Not Eligible')
```

## 3.IF-ELIF-ELSE

```python
age = 20
if age < 14:
    print('Go to Home')
elif age>=14 and age<=18:
    print('Enter the juniors club')
else:
    print('Enter the Adults club')
```
